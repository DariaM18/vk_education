```python
def my_sqrt(x: int) -> int:
    """Возвращает целую часть квадратного корня числа x (округление вниз)."""
    if x < 2:
        return x
    
    left, right = 1, x // 2
    
    while left <= right:
        mid = left + (right - left) // 2
        square = mid * mid
        
        if square == x:
            return mid
        elif square < x:
            left = mid + 1
        else:
            right = mid - 1
            
    # Когда цикл завершается, right указывает на наибольшее число, квадрат которого меньше или равен x
    return right

print(f"Корень из 8: {my_sqrt(8)}")   # Ожидается 2 (2*2=4, 3*3=9)
print(f"Корень из 16: {my_sqrt(16)}") # Ожидается 4
```

```python
def min_copy_time(n: int, x: int, y: int) -> int:
    """Возвращает минимальное время в секундах для создания n копий."""
    if n == 0:
        return 0
    if n == 1:
        return min(x, y)

    # Первую копию делаем на самом быстром ксероксе, чтобы запустить оба
    first_copy_time = min(x, y)
    # Остается сделать n - 1 копий
    remaining_copies = n - 1
    
    # Границы бинарного поиска по времени
    left = 0
    right = remaining_copies * max(x, y) # Худший случай: все копии на медленном
    
    while left < right:
        mid = (left + right) // 2
        # Сколько копий мы успеем сделать за время mid на обоих ксероксах
        copies_made = (mid // x) + (mid // y)
        
        if copies_made >= remaining_copies:
            right = mid # Возможно, мы можем сделать это быстрее
        else:
            left = mid + 1 # Нам нужно больше времени
            
    return first_copy_time + left

print(f"Копии (n=5, x=1, y=2): {min_copy_time(5, 1, 2)} сек.") # Ожидается 4
print(f"Копии (n=4, x=1, y=1): {min_copy_time(4, 1, 1)} сек.") # Ожидается 3
```

```python
def feed_animals(animals: List[int], food: List[int]) -> int:
    """Возвращает максимальное количество накормленных животных."""
    if not animals or not food:
        return 0
    
    animals.sort()
    food.sort()
    
    count_fed = 0
    
    for portion in food:
        # Проверяем, хватит ли текущей порции самому неприхотливому из оставшихся животных
        if portion >= animals[count_fed]:
            count_fed += 1
            
        # Если всех животных накормили, дальше идти нет смысла
        if count_fed == len(animals):
            break
            
    return count_fed

print(f"Накормлено: {feed_animals([3, 4, 7], [8, 1, 2])}")       # Ожидается 1
print(f"Накормлено: {feed_animals([1, 2, 2], [7, 1])}")          # Ожидается 2
print(f"Накормлено: {feed_animals([8, 2, 3, 2], [1, 4, 3, 8])}") # Ожидается 3
```

```python
def find_the_difference(s: str, t: str) -> str:
    """Находит лишнюю букву в строке t, используя XOR."""
    xor_result = 0
    # XOR всех символов первой строки
    for char in s:
        xor_result ^= ord(char)
    # XOR всех символов второй строки
    for char in t:
        xor_result ^= ord(char)
    # Оставшееся число - код лишней буквы
    return chr(xor_result)

print(f"Разница между 'abcd' и 'abcde': {find_the_difference('abcd', 'abcde')}") # Ожидается 'e'
print(f"Разница между '', 'y': {find_the_difference('', 'y')}")                   # Ожидается 'y'
```

```python
def two_sum_unsorted(nums: List[int], target: int) -> List[int]:
    """Находит индексы двух элементов, сумма которых равна target."""
    # Словарь для хранения просмотренных элементов: {значение: индекс}
    seen = {}
    
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        # Если не нашли, сохраняем текущее число и его индекс
        seen[num] = i
        
    return []

print(f"Индексы пары для 9 в [2, 7, 11, 15]: {two_sum_unsorted([2, 7, 11, 15], 9)}") # [0, 1]
print(f"Индексы пары для 6 в [3, 2, 4]: {two_sum_unsorted([3, 2, 4], 6)}")           # [1, 2]
```

```python
def group_anagrams(strs: List[str]) -> List[List[str]]:
    """Группирует анаграммы вместе."""
    anagram_map = {}
    
    for s in strs:
        # Сортируем буквы в строке, чтобы получить уникальный ключ для анаграмм
        sorted_key = "".join(sorted(s))
        
        if sorted_key not in anagram_map:
            anagram_map[sorted_key] = []
        
        anagram_map[sorted_key].append(s)
        
    # Возвращаем только значения словаря (список групп)
    return list(anagram_map.values())

# Тест
input_strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
output = group_anagrams(input_strs)
print(f"Группы анаграмм: {output}")
# Ожидается: [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]
```
