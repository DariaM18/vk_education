```python
from typing import List
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def print_tree(root):
    if not root:
        return "Empty"
    queue = deque([root])
    result = []
    while queue:
        node = queue.popleft()
        if node:
            result.append(str(node.val))
            queue.append(node.left)
            queue.append(node.right)
        else:
            result.append("null")
    while result and result[-1] == "null":
        result.pop()
    print(result)
```

# Задача 1.Восстановление бинарного дерева из массива. 
Дан массив, представляющий бинарное дерево (например, в виде уровнями —
level-order).
Необходимо восстановить структуру бинарного дерева.
# Что нужно сделать:
Реализуйте функцию, которая строит дерево по переданному массиву. Объясните, как
определяется связь родитель–потомок и какая структура данных используется для
хранения узлов.

Объяснение: Дерево задается в виде массива (level-order). Связь родителя и потомка определяется математически по индексам: если родитель под индексом i, то левый ребенок под 2 * i + 1, а правый под 2 * i + 2. Для хранения используем рекурсию.

```python
def build_tree(arr: List[int], i: int = 0) -> TreeNode:
    if i >= len(arr) or arr[i] is None:
        return None
    
    # Создаем текущий узел
    root = TreeNode(arr[i])
    
    # Рекурсивно строим левое и правое поддеревья
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)
    
    return root

# Тест
print("Восстановление дерева:")
arr = [8, 9, 11, 7, 16, 3, 1]
tree = build_tree(arr)
print("Массив:", arr)
print("Дерево (Level-order):")
print_tree(tree) 
```

# Задача 2. Симметричное бинарное дерево
Дано бинарное дерево.
Необходимо определить, является ли оно симметричным относительно своей
середины.
# Что нужно сделать:
Реализуйте рекурсивную или итеративную проверку зеркальности левого и правого
поддеревьев. Поясните, какие пары узлов сравниваются и почему алгоритм работает
корректно.

Объяснение: Используем рекурсию (DFS). Дерево симметрично, если левое поддерево является зеркальным отражением правого. Мы передаем в функцию два узла (левый и правый) и проверяем: их значения равны, и левый ребенок первого равен правому ребенку второго (и наоборот).

```python
def is_symmetric(root: TreeNode) -> bool:
    if not root:
        return True
    
    def is_mirror(left: TreeNode, right: TreeNode) -> bool:
        # Если оба пусты - симметричны
        if not left and not right:
            return True
        # Если один пуст, а другой нет - несимметричны
        if not left or not right:
            return False
        # Значения должны совпадать, и поддеревья должны быть зеркальны
        return (left.val == right.val) and \
               is_mirror(left.left, right.right) and \
               is_mirror(left.right, right.left)
               
    return is_mirror(root.left, root.right)

# Тест
print("Симметричное дерево:")
sym_tree = build_tree([3, 8, 8, 9, 6, 6, 9])
asym_tree = build_tree([3, 7, 8, 9, None, 6, 9])
print(f"Дерево [3, 8, 8, 9, 6, 6, 9] симметрично? {is_symmetric(sym_tree)}")   # True
print(f"Дерево [3, 7, 8, 9, None, 6, 9] симметрично? {is_symmetric(asym_tree)}") # False
```

# Задача 3. Поиск минимальной глубины бинарного дерева
Дано бинарное дерево.
Необходимо найти его минимальную глубину (расстояние от корня до ближайшего
листа).
# Что нужно сделать:
Реализуйте решение с использованием BFS или DFS. Объясните, почему поиск в
ширину (BFS) позволяет корректно найти минимальную глубину.

Объяснение: Минимальная глубина — это путь до ближайшего листа. Лучше всего использовать BFS (поиск в ширину), так как мы обходим дерево по уровням сверху вниз. Первый лист, который мы встретим, будет гарантированно на минимальной глубине.

```python
def min_depth(root: TreeNode) -> int:
    if not root:
        return 0
    
    queue = deque([(root, 1)])  # Храним пару (узел, текущая_глубина)
    
    while queue:
        node, depth = queue.popleft()
        
        # Если это лист (нет детей) - возвращаем текущую глубину
        if not node.left and not node.right:
            return depth
            
        if node.left:
            queue.append((node.left, depth + 1))
        if node.right:
            queue.append((node.right, depth + 1))

# Тест
print("Минимальная глубина:")
tree = build_tree([11, 8, 16, 2, 9, None, 6, 7, None, None, 9])
print_tree(tree)
print(f"Минимальная глубина: {min_depth(tree)}") # Ожидается 3
```

# Задача 4. Поиск произведения максимального и минимального элементов
Дано бинарное дерево поиска (BST), представленное в виде массива.
Необходимо найти произведение минимального и максимального значений.
# Что нужно сделать:
Восстановите BST (если требуется) и найдите минимальный и максимальный
элементы, используя свойства BST. Поясните, почему поиск крайних элементов
выполняется за O(h), где h — высота дерева.

Объяснение: Бинарное дерево поиска (BST) обладает свойством: слева всегда меньшие элементы, справа — большие. Поэтому минимальный элемент — это самый левый узел (идем только по left), а максимальный — самый правый (идем только по right). Сложность O(h), где h — высота дерева. Чтобы найти минимум, нам нужно спуститься от корня как можно ниже влево. Мы делаем ровно один шаг на каждом уровне дерева. Если высота дерева равна h, мы сделаем h шагов. Поэтому сложность — O(h). Чтобы найти максимум, мы аналогично спускаемся только вправо. Это тоже займет O(h).

```python
def get_max_min_product(root: TreeNode) -> int:
    if not root:
        return -1   
    # Проверяем, что в дереве есть хотя бы два элемента
    if not root.left and not root.right:
        return -1
    # 1. поиск минимума
    # Идем от корня влево, пока есть левый потомок
    current_min_node = root
    while current_min_node.left:
        current_min_node = current_min_node.left
    min_val = current_min_node.val
    # 2. поиск максимума
    # Идем от корня вправо, пока есть правый потомок
    current_max_node = root
    while current_max_node.right:
        current_max_node = current_max_node.right
    max_val = current_max_node.val
    # Возвращаем произведение
    return min_val * max_val
print("Массив для восстановления:")
bst_arr = [16, 12, 18, 11, 15, 17, 21, None, None, None, None, None, None, None, 24]
bst_tree = build_tree(bst_arr)
print_tree(bst_tree)
# Вызовем нашу функцию
result = get_max_min_product(bst_tree)
current = bst_tree
while current.left: current = current.left
print(f"минимум = {current.val}")
current = bst_tree
while current.right: current = current.right
print(f"максимум = {current.val}")
print(f"произведение: {result}") # Ожидается 11 * 24 = 264
```

# Задача 5. Сравнение двух бинарных деревьев
Даны два бинарных дерева.
Необходимо определить, являются ли они идентичными (по структуре и значениям
узлов).
# Что нужно сделать:
Реализуйте рекурсивную проверку соответствия узлов обоих деревьев. Объясните,
какие условия должны выполняться, чтобы деревья считались одинаковыми.

Объяснение: Два дерева считаются одинаковыми, если у них одинаковые корни, а их левые и правые поддеревья попарно идентичны. Решается простой рекурсией.

```python
def is_same_tree(p: TreeNode, q: TreeNode) -> bool:
    # Оба пусты - одинаковы
    if not p and not q:
        return True
    # Один пуст, другой нет - разные
    if not p or not q:
        return False
    # Значения не равны - разные
    if p.val != q.val:
        return False
    # Рекурсивно проверяем левые и правые поддеревья
    return is_same_tree(p.left, q.left) and is_same_tree(p.right, q.right)

# Тест
print("Сравнение двух деревьев:")
tree1 = build_tree([16, 12, 18, 15, 17])
tree2 = build_tree([16, 12, 18, 15, 17])
tree3 = build_tree([16, 11, 18, 15, 21])

print(f"tree1 и tree2 одинаковы? {is_same_tree(tree1, tree2)}") # True
print(f"tree1 и tree3 одинаковы? {is_same_tree(tree1, tree3)}") # False
```
