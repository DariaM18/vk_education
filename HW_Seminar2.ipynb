```python
class ListNode:
    """Вспомогательный класс для узла односвязного списка."""
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def create_linked_list(arr):
    """Вспомогательная функция для создания списка из массива."""
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

def print_linked_list(head):
    values = []
    current = head
    while current:
        values.append(str(current.val))
        current = current.next
    print(" -> ".join(values) + " -> None")
```

```python
def reverse_list(head: ListNode) -> ListNode:
    prev = None
    current = head
    
    while current is not None:
        next_node = current.next  
        current.next = prev      
        prev = current            
        current = next_node  
        
    return prev

# Тест
print("Задача 1. Разворот списка:")
head = create_linked_list([1, 2, 3, 4, 5])
print("До:    ", end="")
print_linked_list(head)
reversed_head = reverse_list(head)
print("После: ", end="")
print_linked_list(reversed_head)
```

```python
def middle_node(head: ListNode) -> ListNode:
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        
    return slow

# Тест
print("Задача 2. Середина списка:")
head = create_linked_list([3, 8, 9, 6, 11, 16, 17])
middle = middle_node(head)
print(f"Середина списка начинается с узла: {middle.val}") # Ожидается 6
```

```python
def remove_elements(head: ListNode, val: int) -> ListNode:
    dummy = ListNode(0)
    dummy.next = head
    current = dummy
    
    while current.next is not None:
        if current.next.val == val:
            current.next = current.next.next 
        else:
            current = current.next
            
    return dummy.next

# Тест
print("Задача 3. Удаление элемента (val=3):")
head = create_linked_list([1, 2, 3, 4, 3, 5])
print("До:    ", end="")
print_linked_list(head)
new_head = remove_elements(head, 3)
print("После: ", end="")
print_linked_list(new_head) # Ожидается 1 -> 2 -> 4 -> 5
```

```python
def is_subsequence(a: str, b: str) -> bool:
    i, j = 0, 0
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1
        j += 1
        
    return i == len(a)

# Тест
print("Задача 4. Является ли строка исходной:")
print(f"'abd' в 'abcd': {is_subsequence('abd', 'abcd')}") # True
print(f"'axc' в 'abcd': {is_subsequence('axc', 'abcd')}") # False
```

```python
def two_sum_sorted(nums: list, target: int) -> bool:
    left, right = 0, len(nums) - 1
    
    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
            
    return False

# Тест
print("Задача 5. Поиск пары (Two Sum):")
print(f"В [1, 2, 3, 4, 6] есть пара для 6: {two_sum_sorted([1, 2, 3, 4, 6], 6)}") # True (2+4)
print(f"В [1, 2, 3, 4, 6] есть пара для 11: {two_sum_sorted([1, 2, 3, 4, 6], 11)}") # False
```

```python
def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
        
    return True

# Тест
print("Задача 6. Палиндром:")
print(f"'abcba': {is_palindrome('abcba')}") # True
print(f"'abca': {is_palindrome('abca')}")   # False
```

```python
def remove_duplicates(nums: list) -> int:
    if not nums:
        return 0
        
    slow = 0
    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
            
    return slow + 1

# Тест
print("Задача 7. Удаление дубликатов из массива:")
nums = [1, 1, 2, 2, 3, 4, 4, 5]
print(f"До: {nums}")
length = remove_duplicates(nums)
print(f"После: {nums[:length]} (Длина: {length})")
```

```python
def merge_two_lists(list1: ListNode, list2: ListNode) -> ListNode:
    dummy = ListNode(0)
    current = dummy
    while list1 is not None and list2 is not None:
        if list1.val <= list2.val:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next
        current = current.next 
    if list1 is not None:
        current.next = list1
    elif list2 is not None:
        current.next = list2
        
    return dummy.next

# Тест
print("Задача 8. Слияние двух отсортированных списков:")
l1 = create_linked_list([3, 6, 8])
l2 = create_linked_list([4, 7, 9, 11])
print("Список 1: ", end="")
print_linked_list(l1)
print("Список 2: ", end="")
print_linked_list(l2)
merged = merge_two_lists(l1, l2)
print("Результат:", end="")
print_linked_list(merged)
```
