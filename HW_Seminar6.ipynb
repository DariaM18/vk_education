# Задача 1. Наибольшая непрерывная возрастающая последовательность
Дан массив не отсортированных чисел. Необходимо найти максимально длинную
непрерывную возрастающую последовательность и вернуть её длину.
Примеры:
[3, 2, 8, 9, 5, 10] — ответ 3, так как максимально длинная последовательность: 2, 8, 9
[1, 2, 7, 9, 0, 10] — ответ 4, так как максимально длинная последовательность: 1, 2, 7,
9
[8, 8, 8, 8] — ответ 1
Проанализируйте решение за O(n) и объясните, как отслеживается текущая и
максимальная длина последовательности

Объяснение: Нам нужно найти самый длинный отрезок, где каждое следующее число больше предыдущего. Используем один проход (цикл) за O(n). Мы храним текущую длину (current_len) и максимальную (max_len). Если текущее число больше предыдущего, увеличиваем current_len. Если нет (последовательность прервалась), сбрасываем current_len до 1

```python
def find_length_of_lis(nums: list[int]) -> int:
    if not nums:
        return 0
        
    max_len = 1
    current_len = 1
    
    # Идем по массиву со второго элемента
    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            # Последовательность продолжается
            current_len += 1
            # Обновляем максимум, если текущая длина стала больше
            max_len = max(max_len, current_len)
        else:
            # Последовательность прервалась, начинаем новую с текущего элемента
            current_len = 1
            
    return max_len

# Тесты
print("Непрерывная возрастающая последовательность:")
print(f"[3, 2, 8, 9, 5, 10] -> {find_length_of_lis([3, 2, 8, 9, 5, 10])}") # Ожидается 3 (2, 8, 9)
print(f"[1, 2, 7, 9, 0, 10] -> {find_length_of_lis([1, 2, 7, 9, 0, 10])}") # Ожидается 4 (1, 2, 7, 9)
print(f"[8, 8, 8, 8] -> {find_length_of_lis([8, 8, 8, 8])}")             # Ожидается 1
```

# Задача 2. Треугольник Паскаля
Дано число n. Необходимо построить треугольник Паскаля из n строк.
Объясните, как формируется каждая строка на основе предыдущей и какова
сложность алгоритма.

Объяснение: Каждое число в треугольнике (кроме краев) равно сумме двух чисел над ним. Первая и последняя цифры в каждой строке всегда равны 1. Мы создаем новую строку на основе предыдущей: берем соседние элементы предыдущей строки, складываем их и добавляем в новую, обрамляя единицами.

```python
def generate_pascal_triangle(num_rows: int) -> list[list[int]]:
    if num_rows == 0:
        return []
        
    triangle = [[1]] # Первая строка всегда [1]
    
    for i in range(1, num_rows):
        prev_row = triangle[-1]
        new_row = [1] # Начинаем с 1
        
        # Складываем соседние элементы предыдущей строки
        for j in range(1, len(prev_row)):
            new_row.append(prev_row[j - 1] + prev_row[j])
            
        new_row.append(1) # Заканчиваем 1
        triangle.append(new_row)
        
    return triangle

# Тест
print("Треугольник Паскаля (n=5):")
for row in generate_pascal_triangle(5):
    print(row)
```

# Задача 3. Размен монет
Дан массив номиналов монет coins и число amount. Требуется найти минимальное
количество монет, составляющих сумму amount, либо вернуть -1, если это
невозможно.
Реализуйте решение с использованием динамического программирования и
обоснуйте корректность переходов.

Объяснение: Это классическая задача на Динамическое программирование (DP). Мы создаем массив dp размером amount + 1, где dp[i] — минимальное количество монет для суммы i. Изначально заполняем его бесконечностями (кроме dp[0] = 0). Для каждой суммы от 1 до amount проверяем все монеты. Если монета меньше или равна текущей сумме, мы можем обновить минимум: dp[i] = min(dp[i], dp[i - coin] + 1).

```python
def coin_change(coins: list[int], amount: int) -> int:
    # dp[i] будет хранить минимальное число монет для суммы i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0 # Для суммы 0 нужно 0 монет
    
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                # Либо оставляем текущее значение, либо берем монету + лучший результат для остатка
                dp[i] = min(dp[i], dp[i - coin] + 1)
                
    # Если значение не изменилось, значит сумму собрать нельзя
    return dp[amount] if dp[amount] != float('inf') else -1

# Тесты
print("Размен монет:")
print(f"Монеты [1, 2, 5], сумма 11 -> {coin_change([1, 2, 5], 11)}") # Ожидается 3 (5+5+1)
print(f"Монеты [2], сумма 3 -> {coin_change([2], 3)}")             # Ожидается -1
```

# Задача 4. Максимальный палиндром в строке
Дана строка s. Необходимо найти её самую длинную подстроку-палиндром.
Рассмотрите подход с динамическим программированием или расширением от центра
и проанализируйте сложность алгоритма.

Объяснение: Наиболее оптимальный и простой в написании подход — расширение от центра. Палиндром может быть нечетной длины (центр — один символ) или четной (центр — два одинаковых символа). Мы перебираем каждый символ (и промежуток между символами) как потенциальный центр и расширяемся влево и вправо, пока символы совпадают. Запоминаем самый длинный найденный палиндром. Время O(n^2).

```python
def longest_palindrome(s: str) -> str:
    if not s:
        return ""
    
    start, end = 0, 0
    
    def expand_around_center(left: int, right: int) -> int:
        """Расширяется от центра, возвращает длину палиндрома."""
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        # Возвращаем длину (right - left - 1)
        return right - left - 1

    for i in range(len(s)):
        # Нечетный палиндром (центр в символе i)
        len1 = expand_around_center(i, i)
        # Четный палиндром (центр между i и i+1)
        len2 = expand_around_center(i, i + 1)
        
        max_len = max(len1, len2)
        
        # Если нашли палиндром длиннее, чем был, обновляем границы
        if max_len > end - start:
            # Вычисляем новые начальную и конечную позиции
            start = i - (max_len - 1) // 2
            end = i + max_len // 2
            
    return s[start:end + 1]

# Тесты
print("Самый длинный палиндром:")
print(f"'babad' -> '{longest_palindrome('babad')}'") # Ожидается 'bab' или 'aba'
print(f"'cbbd' -> '{longest_palindrome('cbbd')}'")   # Ожидается 'bb'
```
